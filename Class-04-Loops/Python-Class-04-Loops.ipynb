## Range

`02_range.py`

range: start is included, stop is excluded, step is optional.


```python
print("range(5):")
for i in range(5):
    print(i)

print("range(1, 6):")
for token in range(1, 6):
    print("Token", token)

print("range(2, 11, 2):")
for i in range(2, 11, 2):
    print(i)

print("range(10, 0, -1):")
for sec in range(10, 0, -1):
    print("OTP expires in", sec)

print("range(10, 0) with no step prints nothing:")
for sec in range(10, 0):
    print("OTP expires in", sec)
```

## While

`05_while.py`

while: repeat while the condition is True.
Set the variable before the loop. Change it inside, or the loop never ends.


```python
count = 1

while count <= 5:
    print(count)
    count += 1

# The condition decides how many times, not a number.
balance = 1000

while balance >= 300:
    balance -= 300
    print("Paid ₹300. Left:", balance)
```

## Break continue pass

`07_break_continue_pass.py`

break leaves the loop. continue skips one pass. pass does nothing.


```python
print("break:")
for txn in range(1, 11):
    if txn == 6:
        break
    print("Transaction", txn, "done")

print("continue:")
for txn in range(1, 6):
    if txn == 3:
        continue
    print("Transaction", txn, "done")

print("pass:")
for i in range(5):
    if i == 3:
        pass
    else:
        print(i)
```

## Nested

`08_nested.py`

A loop inside a loop: the inner loop runs fully for each outer pass.


```python
for branch in range(3):
    for counter in range(2):
        print(branch, counter)

# Nested loops draw grids. end=" " keeps the stars on one line.
for row in range(3):
    for column in range(3):
        print("*", end=" ")
    print()

# The inner range may use the outer variable: a growing triangle.
for row in range(1, 4):
    for column in range(row):
        print("₹", end=" ")
    print()
```

## Patterns

`09_patterns.py`

Four loop patterns: accumulate, count, find, filter.
Same skeleton every time: loop, check, update. Start the variable above the loop.


```python
total = 0
for i in range(1, 6):
    total += i
print("Total 1 to 5:", total)

amounts = [250, 1200, 80, 5000, 999, 1500]

# count the big ones
count = 0
for amount in amounts:
    if amount >= 1000:
        count += 1
print("Big payments:", count)

# find: is there a payment of exactly 5000?
found = False
for amount in amounts:
    if amount == 5000:
        found = True
        break
print("Found 5000:", found)

# filter: print only the big ones
for amount in amounts:
    if amount < 1000:
        continue
    print(amount)
```

## Number analyzer

`10_number_analyzer.py`

Mini project: print 1 up to n and label each number even or odd.
range stops before its end value, so the stop is n + 1.


```python
n = int(input("Enter a number: "))

for i in range(1, n + 1):
    if i % 2 == 0:
        print(i, "- Even")
    else:
        print(i, "- Odd")
```

## Paisapay v4

`paisapay_v4.py`

PaisaPay v4: the app keeps running
Built in Class 4. Adds loops to v3: a menu that repeats until Exit,
three PIN attempts, and a mini statement (count, total, biggest).


```python
print("Welcome to PaisaPay")

balance = 1000
saved_pin = "4321"
count = 0
total_spent = 0
biggest = 0

choice = ""
while choice != "4":
    print()
    print("1. Pay   2. Balance   3. Mini statement   4. Exit")
    choice = input("Choose: ")

    if choice == "1":
        payee = input("Pay to: ")
        amount = int(input("Amount in rupees: "))

        pin_ok = False
        for attempt in range(3):
            pin = input("UPI PIN: ")
            if pin == saved_pin:
                pin_ok = True
                break
            print("Wrong PIN.", 2 - attempt, "attempts left")

        if not pin_ok:
            print("Three wrong PINs. Payment refused.")
        elif amount > balance:
            print("Insufficient balance. Payment refused.")
        else:
            balance -= amount
            print("Paid ₹", amount, "to", payee, "| Left ₹", balance)
            count += 1
            total_spent += amount
            if amount > biggest:
                biggest = amount

    elif choice == "2":
        print("Balance: ₹", balance)

    elif choice == "3":
        print("Count:", count, "| Spent ₹", total_spent)
        print("Biggest ₹", biggest)

    elif choice != "4":
        print("Unknown choice. Try again.")

print("Thank you for using PaisaPay")

# What v4 cannot do yet:
#   - fee, cashback and payment modes from v3 are out; they return in v5 as functions
#   - the statement knows totals, not every line (needs a growing list, Class 6)
#   - the same 20 lines repeat for every payment; naming a block once is Class 5
```
