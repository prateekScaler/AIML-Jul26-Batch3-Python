## Index slice

`01_index_slice.py`

A string is a sequence of characters. Index and slice it exactly like a list.


```python
amounts = [600, 300, 1500, 250]
print(amounts[0:3])      # a list slice, Class 6: start included, stop excluded

word = "Python"
print(len(word))

print(word[0])
print(word[3])
print(word[-1])
print(word[-2])

print(word[0:3])
print(word[:3])
print(word[2:])
print(word[::2])
print(word[::-1])
print(word[-4:])

card = "4111222233334444"
print(card[-4:], card[:4])
```

## Length

`02_length.py`

len counts every character, spaces included. The empty string has length 0.


```python
name = "Rashi"
message = "Hello World"

print(len(name))
print(len(message))
print(len(""))
```

## Case strip

`03_case_strip.py`

lower, upper, title, strip: each gives back a new string. Chain them.


```python
text = "Hello Python"
print(text.lower())
print(text.upper())

name = "   rashi sharma   "
print(name.title())
print(name.strip())
print(name.strip().title())

mode = "Card"
print(mode.lower() == "card", mode == "card")
```

## Replace find

`04_replace_find.py`

replace changes every match. find gives the start index, or -1 when not found.


```python
message = "I like Java"
message = message.replace("Java", "Python")
print(message)

text = "I love Python"
print(text.find("Python"))
print(text.find("Java"))

print("banana".replace("a", "o"))
```

## Checks

## Split join

`06_split_join.py`

split cuts a string into a list. join glues a list into a string.


```python
sentence = "Python is easy"
words = sentence.split()
print(words)

data = "apple,banana,mango"
print(data.split(","))

print(" ".join(["Python", "is", "fun"]))
print(", ".join(["Apple", "Banana", "Mango"]))

upi_id = "rashi@okaxis"
parts = upi_id.split("@")
print(len(parts), parts[1])
```

## Fstrings

`07_fstrings.py`

f-strings: values and expressions inside the braces. No str() needed.


```python
name = "Rashi"
age = 20

print("My name is " + name + " and I am " + str(age) + " years old.")
print(f"My name is {name} and I am {age} years old.")

price = 99.5678
quantity = 3
print(f"Total: {price * quantity}")
print(f"Price: {price:.2f}")

amount = 600
fee = 5
print(f"Total ₹{amount + fee}")
print("Hello {name}")        # without the f, the braces are just text
```

## Immutable

`09_immutable.py`

Strings cannot be changed in place. Every string method gives back a new string: catch it.


```python
word = "Python"
# word[0] = "J"               # TypeError: 'str' object does not support item assignment
word = "J" + word[1:]
print(word)

text = "hello"
text.upper()                   # a new string, thrown away
print(text)

text = text.upper()            # caught
print(text)
```

## Iterate vowels

`10_iterate_vowels.py`

Walk through a string one character at a time. Count the vowels.


```python
word = "Python"
for character in word:
    print(character)

word = "education"
count = 0
for character in word:
    if character in "aeiou":
        count += 1
print("Vowels:", count)
```

## Paisapay v7

`paisapay_v7.py`

PaisaPay v7: text in, text out
Built in Class 7. v6 plus strings: every input is cleaned with strip and lower,
a UPI ID is checked with split, a card number is masked with a slice, and the
receipt and messages are f-strings.


```python
print("Welcome to PaisaPay")

saved_pin = "4321"
history = []

def clean(text):
    return text.strip().lower()

def is_valid_amount(amount):
    return amount > 0 and amount <= 100000

def is_valid_pin(pin):
    return pin == saved_pin

def is_valid_upi_id(upi_id):
    parts = upi_id.split("@")
    return len(parts) == 2 and len(parts[0]) > 0 and len(parts[1]) > 0

def mask(card):
    return "**** **** **** " + card[-4:]

def fee_for(mode):
    if mode == "card":
        return 5
    elif mode == "netbanking":
        return 2
    return 0

def cashback_for(amount):
    if amount >= 5000:
        return 100
    elif amount >= 2000:
        return 50
    elif amount >= 500:
        return 10
    return 0

def mode_name(mode):
    match mode:
        case "upi":
            return "UPI"
        case "card":
            return "Card"
        case "netbanking":
            return "Net banking"
        case _:
            return "Unknown mode"

def print_receipt(name, amount, mode, fee, cashback, balance):
    print(f"Paid ₹{amount} to {name} via {mode_name(mode)}")
    print(f"Fee ₹{fee} | Cashback ₹{cashback} | Balance ₹{balance}")

def ask_pin():
    for attempt in range(3):
        pin = input("UPI PIN: ").strip()
        if is_valid_pin(pin):
            return True
        print(f"Wrong PIN. {2 - attempt} attempts left")
    return False

def pay(balance, history, upi_id, amount, mode="upi"):
    if not is_valid_upi_id(upi_id):
        print(f"'{upi_id}' is not a valid UPI ID. Payment refused.")
        return balance
    if not is_valid_amount(amount):
        print("Invalid amount. Payment refused.")
        return balance
    if mode_name(mode) == "Unknown mode":
        print(f"Unknown payment mode '{mode}'. Payment refused.")
        return balance
    fee = fee_for(mode)
    if amount + fee > balance:
        print("Insufficient balance. Payment refused.")
        return balance
    cashback = cashback_for(amount)
    balance = balance - amount - fee + cashback
    name = upi_id.split("@")[0].title()
    print_receipt(name, amount, mode, fee, cashback, balance)
    history.append(amount)
    return balance

def mini_statement(history):
    if len(history) == 0:
        print("No payments yet.")
        return
    total = 0
    for amount in history:
        total += amount
    print(f"Payments: {len(history)} | Spent ₹{total}")
    print(f"Biggest ₹{sorted(history)[-1]} | Last five: {history[-5:]}")

balance = 1000
choice = ""
while choice != "4":
    print()
    print("1 Pay  2 Balance  3 Statement  4 Exit  5 Add  6 Card")
    choice = clean(input("Choose: "))

    if choice == "1":
        upi_id = clean(input("Pay to (UPI ID): "))
        amount = int(input("Amount in rupees: ").strip())
        mode = clean(input("Mode (upi / card / netbanking): "))
        if ask_pin():
            balance = pay(balance, history, upi_id, amount, mode)
        else:
            print("Three wrong PINs. Payment refused.")

    elif choice == "2":
        print(f"Balance: ₹{balance}")

    elif choice == "3":
        mini_statement(history)

    elif choice == "5":
        amount = int(input("Amount to add: ").strip())
        if is_valid_amount(amount):
            balance = balance + amount
            print(f"Added ₹{amount}. Balance ₹{balance}")
        else:
            print("Invalid amount. Nothing added.")

    elif choice == "6":
        card = input("Card number: ").replace(" ", "").strip()
        if len(card) == 16:
            print(f"Card linked: {mask(card)}")
        else:
            print("A card number has 16 digits.")

    elif choice != "4":
        print(f"Unknown choice '{choice}'. Try again.")

print("Thank you for using PaisaPay")

# What v7 cannot do yet:
#   - keep the name, amount and mode of one payment together (records, Class 8)
#   - survive a non-number typed as an amount (errors, Class 12)
#   - check a UPI ID or a card number properly (regex, Class 11)
```
