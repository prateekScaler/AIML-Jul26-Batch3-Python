```python
name = "Kishor"      # str   - text, inside quotes
age = 25              # int   - whole number
height = 5.8          # float - number with a decimal point
is_student = True     # bool  - True or False (capital first letter)

print(name)
print(age)
print(height)
print(is_student)
```

## Assignment

`04_assignment.py`

= means "assign", not "is equal to".
Python works out the RIGHT side first, then stores the result under the name.


```python
age = 20
print(age)        # 20

age = age + 1     # right side: 20 + 1 = 21, then store 21 under age
print(age)        # 21

# A wallet balance changes the same way:
balance = 1000
balance = balance - 200
balance = balance - 300
print(balance)    # 500

# Assignment copies the value. It does not link two names together.
price = 499
total = price
price = 599
print(total)      # still 499
```

## Check types

`05_check_types.py`

type() answers: what kind of data is stored here?


```python
age = 25
height = 5.8
name = "Kishor"
is_student = True

print(type(age))          # <class 'int'>
print(type(height))       # <class 'float'>
print(type(name))         # <class 'str'>
print(type(is_student))   # <class 'bool'>

# Quotes and the decimal point decide the type, not how the value looks:
print(type(42))       # <class 'int'>
print(type(42.0))     # <class 'float'>
print(type("42"))     # <class 'str'>
print(type("True"))   # <class 'str'>
```

## Input

`07_input.py`

input() pauses the program, shows a prompt, and waits for the user.

```python
name = input("What is your name? ")
print(name)

# The big surprise: input() ALWAYS returns a string (str),
# even when the user types digits.
age = input("What is your age? ")
print(age)
print(type(age))    # <class 'str'>
```

## Type casting

`08_type_casting.py`

Type casting = converting a value from one type to another.
The problem: input() gives text, and text + number is not allowed.
age = input("What is your age? ")
next_year = age + 1
TypeError: can only concatenate str (not "int") to str
The fix: convert the text to an int as soon as it arrives.


```python
age = int(input("What is your age? "))
print(type(age))       # <class 'int'>
print(age + 1)         # real arithmetic now works

# The casting functions
print(int("25"))       # 25     text  -> whole number
print(float("99.5"))   # 99.5   text  -> decimal number
print(str(100))        # 100    number -> text (it prints the same, but it is a str)

# Things worth trying one at a time:
#   int("99.5")   -> ValueError: invalid literal for int() with base 10: '99.5'
#   int(99.9)     -> 99  (cuts off the decimal part, does not round)
#   "₹" + 500     -> TypeError;  "₹" + str(500) works
```

## Money and floats

`12_money_and_floats.py`

Why payment systems do not store money as a float.


```python
print(0.1 + 0.2)          # 0.30000000000000004, not 0.3

# Floats are stored in binary, and many decimals cannot be stored exactly.
# So real payment systems keep money as an int in the smallest unit (paise).

price_paise = 29900       # 299 rupees, stored as 29900 paise
fee_paise = 10            # 10 paise
print(price_paise + fee_paise)   # 29910, exact

# Casting a float to an int cuts the decimal part off. It does not round.
print(int(99.9))          # 99
```

## Paisapay v1

`paisapay_v1.py`

PaisaPay v1: take one payment and print a receipt
Built in Class 1 using print(), variables, input() and type casting.


```python
print("Welcome to PaisaPay")

name = input("Your name: ")
payee = input("Pay to: ")
amount = int(input("Amount in rupees: "))

balance = 1000              # every new wallet starts with ₹1000
balance = balance - amount  # the money leaves the wallet

print("----- Receipt -----")
print("From:", name)
print("To:", payee)
print("Paid: ₹", amount)
print("Balance left: ₹", balance)

# What v1 cannot do yet:
#   - refuse a payment bigger than the balance (needs decisions, Class 3)
#   - accept an amount like 99.50 (int() rejects it)
#   - take more than one payment (needs loops, Class 4)
```
