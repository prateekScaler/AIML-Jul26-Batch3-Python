## Read a file

`02_read_a_file.py`

open() asks the operating system for the file and gives back a file object (a handle).
.read() gives the whole file as one string. Reading moves a bookmark forward.
The first lines only make the sample file, so this runs anywhere.


```python
with open("passbook.txt", "w") as f:
    f.write("Paid 600 to Sumit\nPaid 250 to Vamsi\n")

file = open("passbook.txt")
print(file)                    # a file object, not the text

contents = file.read()
print(contents)
print(type(contents))          # <class 'str'>
file.close()

# the bookmark: a second read on the same open file gets nothing
file = open("passbook.txt")
first = file.read()
second = file.read()
print(len(first), len(second))  # 36 0
file.close()
```

## File modes

`03_file_modes.py`

The four modes, each with both cases: the file exists, and the file does not.


```python
def show(name):
    with open(name) as f:
        print(f"{name}: {f.read().splitlines()}")

def fresh_passbook():
    with open("passbook.txt", "w") as f:
        f.write("Paid 600 to Sumit\nPaid 250 to Vamsi\n")

# "r": read only. Writing fails; a missing file fails.
fresh_passbook()
try:
    file = open("passbook.txt", "r")
    file.write("Paid 100 to Akshay")
except Exception as e:
    print("r, write:", type(e).__name__, e)
try:
    open("missing.txt", "r")
except FileNotFoundError as e:
    print("r, missing:", e)

# "w": creates a missing file; erases a file that exists
file = open("receipt.txt", "w")
file.write("Paid 100 to Akshay")
file.close()
show("receipt.txt")
file = open("passbook.txt", "w")
file.write("Paid 100 to Akshay")
file.close()
show("passbook.txt")             # the two old payments are gone

# "a": adds at the end; creates a missing file
fresh_passbook()
file = open("passbook.txt", "a")
file.write("Paid 100 to Akshay\n")
file.close()
show("passbook.txt")

# "x": creates a new file; refuses a file that exists
try:
    open("passbook.txt", "x")
except FileExistsError as e:
    print("x, exists:", e)
```

## With

`04_with.py`

Why close: writes wait in a buffer in RAM until the file is closed.


```python
file = open("receipt.txt", "w")
file.write("Paid 100 to Akshay")
print("[" + open("receipt.txt").read() + "]")   # []: not in storage yet
file.close()
print("[" + open("receipt.txt").read() + "]")   # [Paid 100 to Akshay]

# with opens the file and closes it for you, even if an error happens inside.
with open("passbook.txt", "w") as f:
    f.write("Paid 600 to Sumit\nPaid 250 to Vamsi\n")

with open("passbook.txt", "r") as file:
    contents = file.read()
    print(contents)

print("Is the file closed now?", file.closed)
```

## Line by line

`05_line_by_line.py`

Looping over a file gives one line at a time. Each line keeps its \n.


```python
with open("passbook.txt", "w") as f:
    f.write("Paid 600 to Sumit\nPaid 250 to Vamsi\n")

with open("passbook.txt", "r") as file:
    for line in file:
        print(line)                # two newlines: the line's own, and print's

with open("passbook.txt", "r") as file:
    for line in file:
        print(line.strip())        # strip removes the line's \n

# lines become data: split each line, convert the amount, add it up
total = 0
with open("passbook.txt") as file:
    for line in file:
        words = line.split()
        total += int(words[1])
print("Total spent:", total)
```

## Write and append

`06_write_and_append.py`

write() adds no newline: put \n yourself. "w" inside a loop keeps only the last write.


```python
with open("passbook.txt", "w") as f:
    f.write("Paid 600 to Sumit\nPaid 250 to Vamsi\n")

with open("passbook.txt", "a") as file:
    file.write("Paid 100 to Akshay\n")

with open("passbook.txt") as f:
    print(f.read())

for name in ["Sumit", "Vamsi", "Akshay"]:      # the bug: "w" in a loop
    f = open("paid.txt", "w")
    f.write(name + "\n")
    f.close()
with open("paid.txt") as f:
    print(f.read())                            # only Akshay
```
