## Tuples

`01_tuples.py`

A tuple is a fixed collection: round brackets, and it cannot change.


```python
point = (3, 4)
print(point)
print(point[0])

# The comma makes the tuple, not the brackets.
point = 3, 4
print(point)
print(type(point))

# The single-element trap: (5) is just 5. (5,) is a tuple.
x = (5)
y = (5,)
print(type(x))
print(type(y))

# Index, slice, unpack: the same as a list.
coords = (10, 20, 30, 40)
print(coords[0])
print(coords[1:3])
print(coords[-1])

x, y = (3, 4)
print(x)
print(y)

# Only two methods: count and index. Nothing to add, remove or sort.
coords = (10, 20, 10, 30, 10)
print(coords.count(10))
print(coords.index(30))

# Trying to change a tuple. Remove the # to see the error.
# coords[0] = 99      # TypeError: 'tuple' object does not support item assignment
```

## Why tuples

`02_why_tuples.py`

A function that returns two values returns a tuple.


```python
def min_and_max(numbers):
    return min(numbers), max(numbers)

result = min_and_max([600, 1500, 250])
print(result)
print(type(result))

smallest, biggest = min_and_max([600, 1500, 250])
print(smallest, biggest)

# A payment as a fixed record: name, amount, mode. A receipt should not change.
payment = ("Rajani", 600, "upi")
name, amount, mode = payment
print(f"Paid ₹{amount} to {name} via {mode}")
```

## Sets

`03_sets.py`

A set keeps one copy of each value. Curly brackets, no order.


```python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)
print(len(numbers))

# The empty-set trap: {} is an empty dictionary.
empty = {}
print(type(empty))

empty_set = set()
print(type(empty_set))

# add, remove, discard
fruits = {"apple", "banana"}
fruits.add("cherry")
print(len(fruits), "cherry" in fruits)

fruits.discard("mango")      # not there: nothing happens
print(len(fruits))

# fruits.remove("mango")     # not there: KeyError: 'mango'
```

## Set operations

`04_set_operations.py`

Set operations: the same ideas as groups in maths.


```python
python_students = {"Praveen", "Rahul", "Vamsi"}
sql_students = {"Rahul", "Vamsi", "Akshay"}

print(python_students | sql_students)    # union: in either
print(python_students & sql_students)    # intersection: in both
print(python_students - sql_students)    # difference: only in Python
print(python_students ^ sql_students)    # symmetric difference: in exactly one

# Membership
print("Rahul" in python_students)
print("Akshay" in python_students)
```

## Dicts

`05_dicts.py`

A dictionary looks things up by name, not by position.


```python
student = {"name": "Rajani", "age": 20, "course": "AIML"}
print(student["name"])
print(student["age"])

# Update an existing key, add a new key.
student["age"] = 21
student["city"] = "Pune"
print(student)

# get: safe lookup. Square brackets: KeyError when the key is missing.
print(student.get("age"))
print(student.get("grade"))
print(student.get("grade", "Not assigned"))
# print(student["grade"])    # KeyError: 'grade'
```

## Dict methods

`06_dict_methods.py`


```python
student = {"name": "Rajani", "age": 20}

print(student.keys())
print(student.values())
print(student.items())

student.update({"age": 22, "grade": "A"})
print(student)

removed = student.pop("grade")
print(removed)
print(student)
```

## Overwrite

`07_overwrite.py`

A key points to one value. Assign again and the old value is gone.


```python
scores = {}
scores["Aanand"] = 90
scores["Saurabh"] = 85
scores["Aanand"] = 95

print(scores)
print(len(scores))

# To keep every score, store a list under the key.
scores = {}
scores["Aanand"] = [90]
scores["Saurabh"] = [85]
scores["Aanand"].append(95)
print(scores)
```

## Iterate

`08_iterate.py`


```python
student = {"name": "Rajani", "age": 22, "course": "AIML"}

# A loop over a dictionary gives the keys.
for key in student:
    print(key)

# items() gives key and value together.
for key, value in student.items():
    print(key, "->", value)

# values() gives only the values.
for value in student.values():
    print(value)

# A tuple keeps its order. A set makes no promise about order.
coords = (10, 20, 30)
for value in coords:
    print(value)

fruits = {"apple", "banana", "cherry"}
for fruit in fruits:
    print(fruit)
```

## Nested

`09_nested.py`

A dictionary of lists: every score survives.


```python
scores = {
    "Prasad": [90, 85, 95],
    "Madhumaya": [70, 75, 80]
}
print(scores["Prasad"])
print(scores["Prasad"][0])

# A dictionary of dictionaries: a lookup inside a lookup.
contacts = {
    "sumit@ybl": {"name": "Sumit", "city": "Bengaluru"},
    "parveen@okaxis": {"name": "Parveen", "city": "Pune"}
}
print(contacts["sumit@ybl"]["city"])

# A list of tuples: unpack each record in the for line.
payments = [("Sumit", 600, "upi"),
            ("Parveen", 250, "card"),
            ("Sumit", 1500, "upi")]
for name, amount, mode in payments:
    print(name, amount, mode)
```

## Word counts

`10_word_counts.py`

Mini project: count every word, then count the unique words.


```python
sentence = "the cat sat on the mat the cat ran"
words = sentence.split()
print(words)

counts = {}
for word in words:
    if word in counts:
        counts[word] += 1
    else:
        counts[word] = 1
print(counts)

unique_words = set(words)
print("Unique words:", len(unique_words))
```

## Paisapay v8

`paisapay_v8.py`

PaisaPay v8: real records
Built in Class 8. v7 plus tuples, sets and dictionaries: every payment is a
(name, amount, mode) tuple, the statement counts unique payees with a set,
contacts are a dictionary from UPI ID to name, and spending per payee is a
dictionary built with get.


```python
print("Welcome to PaisaPay")

saved_pin = "4321"
history = []
contacts = {"rajani@okaxis": "Rajani", "sumit@ybl": "Sumit"}

def clean(text):
    return text.strip().lower()

def is_valid_amount(amount):
    return amount > 0 and amount <= 100000

def is_valid_pin(pin):
    return pin == saved_pin

def is_valid_upi_id(upi_id):
    parts = upi_id.split("@")
    return len(parts) == 2 and len(parts[0]) > 0 and len(parts[1]) > 0

def mask(card):
    return "**** **** **** " + card[-4:]

def fee_for(mode):
    if mode == "card":
        return 5
    elif mode == "netbanking":
        return 2
    return 0

def cashback_for(amount):
    if amount >= 5000:
        return 100
    elif amount >= 2000:
        return 50
    elif amount >= 500:
        return 10
    return 0

def mode_name(mode):
    match mode:
        case "upi":
            return "UPI"
        case "card":
            return "Card"
        case "netbanking":
            return "Net banking"
        case _:
            return "Unknown mode"

def name_for(upi_id):
    return contacts.get(upi_id, upi_id.split("@")[0].title())

def print_receipt(name, amount, mode, fee, cashback, balance):
    print(f"Paid ₹{amount} to {name} via {mode_name(mode)}")
    print(f"Fee ₹{fee} | Cashback ₹{cashback} | Balance ₹{balance}")

def ask_pin():
    for attempt in range(3):
        pin = input("UPI PIN: ").strip()
        if is_valid_pin(pin):
            return True
        print(f"Wrong PIN. {2 - attempt} attempts left")
    return False

def pay(balance, history, upi_id, amount, mode="upi"):
    if not is_valid_upi_id(upi_id):
        print(f"'{upi_id}' is not a valid UPI ID. Payment refused.")
        return balance
    if not is_valid_amount(amount):
        print("Invalid amount. Payment refused.")
        return balance
    if mode_name(mode) == "Unknown mode":
        print(f"Unknown payment mode '{mode}'. Payment refused.")
        return balance
    fee = fee_for(mode)
    if amount + fee > balance:
        print("Insufficient balance. Payment refused.")
        return balance
    cashback = cashback_for(amount)
    balance = balance - amount - fee + cashback
    name = name_for(upi_id)
    print_receipt(name, amount, mode, fee, cashback, balance)
    history.append((name, amount, mode))
    return balance

def mini_statement(history):
    if len(history) == 0:
        print("No payments yet.")
        return
    total = 0
    payees = set()
    spent = {}
    for name, amount, mode in history:
        total += amount
        payees.add(name)
        spent[name] = spent.get(name, 0) + amount
    print(f"Payments: {len(history)} | Spent ₹{total}")
    print(f"Payees: {len(payees)}")
    for name, amount in spent.items():
        print(f"  {name}: ₹{amount}")
    print(f"Last: {history[-1]}")

balance = 1000
choice = ""
while choice != "4":
    print()
    print("1 Pay  2 Balance  3 Statement  4 Exit  5 Add  6 Card  7 Save")
    choice = clean(input("Choose: "))

    if choice == "1":
        upi_id = clean(input("Pay to (UPI ID): "))
        amount = int(input("Amount in rupees: ").strip())
        mode = clean(input("Mode (upi / card / netbanking): "))
        if ask_pin():
            balance = pay(balance, history, upi_id, amount, mode)
        else:
            print("Three wrong PINs. Payment refused.")

    elif choice == "2":
        print(f"Balance: ₹{balance}")

    elif choice == "3":
        mini_statement(history)

    elif choice == "5":
        amount = int(input("Amount to add: ").strip())
        if is_valid_amount(amount):
            balance = balance + amount
            print(f"Added ₹{amount}. Balance ₹{balance}")
        else:
            print("Invalid amount. Nothing added.")

    elif choice == "6":
        card = input("Card number: ").replace(" ", "").strip()
        if len(card) == 16:
            print(f"Card linked: {mask(card)}")
        else:
            print("A card number has 16 digits.")

    elif choice == "7":
        upi_id = clean(input("Contact UPI ID: "))
        name = input("Name: ").strip().title()
        if is_valid_upi_id(upi_id):
            contacts[upi_id] = name
            print(f"Saved. {len(contacts)} contacts.")
        else:
            print(f"'{upi_id}' is not a valid UPI ID.")

    elif choice != "4":
        print(f"Unknown choice '{choice}'. Try again.")

print("Thank you for using PaisaPay")

# What v8 cannot do yet:
#   - find the biggest payment or sort payments by amount (map, filter, sorted: Class 9)
#   - keep the rules and the data of an account together (classes, Class 10)
#   - remember anything after the program ends (files, Class 12)
```
