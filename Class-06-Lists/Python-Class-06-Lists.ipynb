## Revision functions

`00_revision_functions.py`

Revision from Class 5: scope, and the order of parameters.
A function can read a global, but assigning inside makes a new local.


```python
x = 10

def show():
    x = 5          # a new local x
    print(x)

show()             # 5
print(x)           # 10: the outside x never changed

# A function cannot change an outside number. It hands back a new one.
balance = 1000

def add_bonus(balance):
    return balance + 100

balance = add_bonus(balance)
print(balance)     # 1100

# Definition: required parameters first, defaults last.
# Call: positional arguments first, keyword arguments last.
def pay(amount, mode="upi", note=""):
    print(amount, mode, note)

pay(500)
pay(500, "card")
pay(500, note="rent")
# pay(mode="card", 500)         # SyntaxError: positional argument follows keyword argument
# def pay(mode="upi", amount)   # SyntaxError: parameter without a default follows parameter with a default
```

## Change alias

`03_change_alias.py`

Lists are mutable: a box can be replaced. And two names can point at one list.


```python
fruits = ["Apple", "Banana", "Mango"]
fruits[1] = "Orange"
print(fruits)

original = [10, 20, 30]
alias = original          # the same list, a second name
alias[0] = 99
print(original)           # [99, 20, 30]

copy = original.copy()    # a real second list
copy[0] = 10
print(original, copy)
```

## Slicing

```python
numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[:3])
print(numbers[2:])
print(numbers[-2:])
print(numbers[:])

numbers = [1, 2, 3, 4, 5, 6]
print(numbers[0:6:2])
print(numbers[1:5:2])
print(numbers[::-1])
```

## Remove pop

## Sort sorted

`07_sort_sorted.py`

sort() rearranges the list in place and gives back None.
sorted() gives back a new sorted list and leaves the original untouched.


```python
numbers = [50, 10, 40, 20, 30]
numbers.sort()
print(numbers)

names = ["Vinay", "Akash", "Shubhodip", "Hitender"]
names.sort()
print(names)

numbers.sort(reverse=True)
print(numbers)

numbers = [50, 10, 40, 20, 30]
sorted_numbers = numbers.sort()
print(sorted_numbers)        # None: sort() hands nothing back
print(numbers)               # but the list is sorted

numbers = [50, 10, 40, 20, 30]
new_list = sorted(numbers)
print(new_list)
print(numbers)
```

## Iterate

`08_iterate.py`

Walk through a list with for. Use range(len(...)) when you also need the position.


```python
students = ["Hitender", "Parvatha", "Shubhodip", "Akash"]

for student in students:
    print(student)

for i in range(len(students)):
    print(i, students[i])

scores = [35, 72, 45, 90, 28]
for score in scores:
    if score >= 40:
        print(score, "Pass")
    else:
        print(score, "Fail")
```

## Comprehension

`09_comprehension.py`

A list comprehension builds a new list in one line: [expression for item in list if condition].


```python
numbers = [1, 2, 3, 4, 5]

squares = []
for number in numbers:
    squares.append(number * number)
print(squares)

squares = [number * number for number in numbers]
print(squares)

numbers = [1, 2, 3, 4, 5, 6]
evens = [number for number in numbers if number % 2 == 0]
print(evens)

numbers = [2, 7, 4, 9, 1, 8]
names = ["asha", "rahul", "priya"]
print([number * 2 for number in numbers])
print([number for number in numbers if number > 5])
print([name.upper() for name in names])
```

## Paisapay v6

`paisapay_v6.py`

PaisaPay v6: the app remembers
Built in Class 6. v5 plus one list, history, and one function that reads it:
mini_statement counts with len, adds with a loop, finds the biggest with sorted,
takes the last five with a slice, and lists the big payments with a comprehension.


```python
print("Welcome to PaisaPay")

saved_pin = "4321"
history = []

def is_valid_amount(amount):
    return amount > 0 and amount <= 100000

def is_valid_pin(pin):
    return pin == saved_pin

def fee_for(mode):
    if mode == "card":
        return 5
    elif mode == "netbanking":
        return 2
    return 0

def cashback_for(amount):
    if amount >= 5000:
        return 100
    elif amount >= 2000:
        return 50
    elif amount >= 500:
        return 10
    return 0

def mode_name(mode):
    match mode:
        case "upi":
            return "UPI"
        case "card":
            return "Card"
        case "netbanking":
            return "Net banking"
        case _:
            return "Unknown mode"

def show_balance(balance):
    print("Balance: ₹", balance)

def add_money(balance, amount):
    if not is_valid_amount(amount):
        print("Invalid amount. Nothing added.")
        return balance
    print("Added ₹", amount)
    return balance + amount

def ask_pin():
    for attempt in range(3):
        pin = input("UPI PIN: ")
        if is_valid_pin(pin):
            return True
        print("Wrong PIN.", 2 - attempt, "attempts left")
    return False

def pay(balance, history, payee, amount, mode="upi"):
    if not is_valid_amount(amount):
        print("Invalid amount. Payment refused.")
        return balance
    if mode_name(mode) == "Unknown mode":
        print("Unknown payment mode. Payment refused.")
        return balance
    fee = fee_for(mode)
    if amount + fee > balance:
        print("Insufficient balance. Payment refused.")
        return balance
    cashback = cashback_for(amount)
    print("Paid ₹", amount, "to", payee, "via", mode_name(mode))
    print("Fee ₹", fee, "| Cashback ₹", cashback)
    history.append(amount)
    return balance - amount - fee + cashback

def mini_statement(history):
    if len(history) == 0:
        print("No payments yet.")
        return
    total = 0
    for amount in history:
        total += amount
    print("Payments:", len(history), "| Spent ₹", total)
    print("Biggest ₹", sorted(history)[-1])
    print("Last five:", history[-5:])
    big = [amount for amount in history if amount >= 500]
    print("Big payments:", big)

balance = 1000
choice = ""
while choice != "4":
    print()
    print("1. Pay  2. Balance  3. Statement  4. Exit  5. Add money")
    choice = input("Choose: ")

    if choice == "1":
        payee = input("Pay to: ")
        amount = int(input("Amount in rupees: "))
        mode = input("Mode (upi / card / netbanking): ")
        if ask_pin():
            balance = pay(balance, history, payee, amount, mode)
        else:
            print("Three wrong PINs. Payment refused.")

    elif choice == "2":
        show_balance(balance)

    elif choice == "3":
        mini_statement(history)

    elif choice == "5":
        amount = int(input("Amount to add: "))
        balance = add_money(balance, amount)

    elif choice != "4":
        print("Unknown choice. Try again.")

print("Thank you for using PaisaPay")

# What v6 cannot do yet:
#   - say who was paid or when: a payment as one record is Class 8
#   - clean typed input or show paise: strings and formatting, Class 7
#   - keep the history after the app closes: files, Class 12
```
