## Type casting

`02_type_casting.py`

Type casting = converting a value from one type to another.
Convert the text as soon as it arrives, then the maths works.


```python
a = int(input("First amount: "))
b = int(input("Second amount: "))

print(a + b)    # 150 when the user types 100 and 50

print(int("25"))       # 25
print(float("99.5"))   # 99.5
print(str(100))        # 100, but it is now text

# Worth trying one at a time:
#   int("99.5")   -> ValueError
#   int(99.9)     -> 99 (cuts the decimals, does not round)
#   "₹" + 500     -> TypeError;  "₹" + str(500) works
```

## Arithmetic

`03_arithmetic.py`

Arithmetic operators


```python
a = 10
b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30   (* not ×)
print(a / b)    # 3.3333333333333335   / always gives a float
print(15 / 5)   # 3.0  even when it divides evenly
print(a // b)   # 3    floor division: the whole share
print(a % b)    # 1    modulus: what is left over
print(2 ** 3)   # 8    power: 2 × 2 × 2

# Predict each one before you run it
print(15 + 5)
print(15 - 5)
print(15 * 5)
print(15 / 5)
print(15 // 4)
print(15 % 4)
print(2 ** 4)
```

## Floor and remainder

`04_floor_and_remainder.py`

// and % are two halves of one division.


```python
print(10 // 3)   # 3  each friend gets ₹3
print(10 % 3)    # 1  ₹1 is left over

# A ₹450 dinner bill, split among 4
print(450 // 4)  # 112
print(450 % 4)   # 2

# % is the standard test for "divides evenly"
print(14 % 2)      # 0 -> even
print(15 % 2)      # 1 -> odd
print(2550 % 100)  # 50 -> an ATM with only ₹100 notes cannot give ₹2,550
```

## Cut or round

`05_cut_or_round.py`

Cutting off decimals is not the same as rounding.
In 1982-83 the Vancouver Stock Exchange index cut instead of rounding,
thousands of times a day, and drifted from 1000 down to 524.811.
Recalculated correctly it was 1098.892.


```python
price = 99.9
print(int(price))     # 99   cuts
print(round(price))   # 100  rounds
```

## Comparison

`06_comparison.py`

Comparison operators ask a question. The answer is always True or False.


```python
print(10 > 5)     # True
print(10 < 5)     # False
print(10 == 10)   # True
print(10 != 10)   # False
print(5 >= 5)     # True
print(3 <= 2)     # False

# The answer can be stored like any other value
age = 20
is_adult = age >= 18
print(is_adult)   # True
```

## Assign vs compare

`07_assign_vs_compare.py`

=  assigns:   put 20 into age
== compares:  is age equal to 20?


```python
age = 20
print(age == 20)    # True

# A str is never equal to an int. No error, just False.
pin = "4321"
print(pin == 4321)      # False
print(pin == "4321")    # True

# What to do instead: compare like with like.
print(pin == "4321")        # True   text with text (a PIN is text; keeps a leading zero)
print(int(pin) == 4321)     # True   number with number, converted before comparing
```

## Logical

`08_logical.py`

Logical operators combine conditions.


```python
age = 20
has_id = True
print(age >= 18 and has_id)    # True   and: both must be True

has_card = False
has_cash = True
print(has_card or has_cash)    # True   or: at least one must be True

is_raining = False
print(not is_raining)          # True   not: flips the value

# Comparisons create conditions. Logical operators combine them.
can_enter = age >= 18 and has_id
print(can_enter)               # True

# Other ways to write the same condition
can_enter = has_id and age >= 18             # same checks, other order
is_adult = age >= 18                         # name each check first
can_enter = is_adult and has_id
can_enter = not (age < 18 or not has_id)     # "not blocked" says the same thing
print(can_enter)                             # True
```

## Assignment shortcuts

## Membership identity

`10_membership_identity.py`

in / not in ask: is this inside that?


```python
upi_id = "rajani@paisapay"
print("@" in upi_id)        # True
print(" " not in upi_id)    # True

# == asks: same value?   is asks: the very same object?
a = [100, 200]
b = [100, 200]
print(a == b)   # True
print(a is b)   # False: two separate lists that happen to hold the same values

# Where is says True: two names for one object
b = a
print(a is b)   # True
b += [300]
print(a)        # [100, 200, 300]  the same list, seen through either name
# The everyday use of is: checking for "nothing", x is None (functions class).
```

## Precedence

`11_precedence.py`

Operator precedence: the order Python follows.
()  then **  then * / // %  then + -  then comparisons  then not, and, or


```python
print(10 + 5 * 2)       # 20  multiplication first
print((10 + 5) * 2)     # 30  parentheses decide
print(2 + 3 ** 2)       # 11  power first

# A real bug: the average of two amounts
print(100 + 200 / 2)    # 200.0  wrong
print((100 + 200) / 2)  # 150.0  right

print(True or False and False)   # True: and goes before or
```

## Float why

`14_float_why.py`

Why floats are not exact.
In decimal, 1/3 never ends. Cut it to 0.33 and the pieces no longer add up to 1.


```python
print(0.33 + 0.33 + 0.33)   # 0.99

# A computer counts in binary, and in binary 0.1 never ends either.
# So it is stored cut short, and the tiny error shows up when you add.
print(0.1 + 0.2)            # 0.30000000000000004
print(0.1 * 3)              # 0.30000000000000004

# This is why payment systems keep money as whole paise, not float rupees.
```

## Paisapay v2

`paisapay_v2.py`

PaisaPay v2: money maths and checks
Built in Class 2. Adds operators to v1: a fee, cashback, a bill split,
and two questions the program can now ask about a payment.


```python
print("Welcome to PaisaPay")

name = input("Your name: ")
payee = input("Pay to: ")
amount = int(input("Amount in rupees: "))
people = int(input("Split between how many people? "))

balance = 1000
fee = 2                          # flat convenience fee, in rupees
total = amount + fee
cashback = amount * 5 // 100     # 5% cashback, in whole rupees
share = amount // people         # each person's share
left_over = amount % people      # rupees that do not divide evenly

can_pay = balance >= total
is_big_payment = amount >= 500

balance -= total
balance += cashback

print("----- Receipt -----")
print("From:", name)
print("To:", payee)
print("Paid: ₹", amount, "+ fee ₹", fee)
print("Cashback: ₹", cashback)
print("Each person owes: ₹", share, "| left over: ₹", left_over)
print("Enough balance:", can_pay)
print("Big payment:", is_big_payment)
print("Balance left: ₹", balance)

# What v2 cannot do yet:
#   - refuse a payment: it prints "Enough balance: False" and pays anyway
#   - act on any of its True/False answers (needs decisions, Class 3)
#   - take more than one payment (needs loops, Class 4)
```
