## Parameters

`02_parameters.py`

Parameters and arguments: the variable in the definition, the value at the call.
With several parameters, arguments are matched by position.


```python
def greet(name):
    print("Hello", name)

greet("Uma")
greet("Prince")
greet("Vinit")

def introduce(name, age):
    print("My name is", name)
    print("I am", age, "years old.")

introduce("Uma", 20)
```

## Return

`03_return.py`

print shows a value to the human. return hands it back to the program.
A function with no return hands back None.


```python
def add_print(a, b):
    print(a + b)

result = add_print(10, 20)
print("stored:", result)          # None: nothing was returned

def add(a, b):
    return a + b

result = add(10, 20)
print("stored:", result)          # 30

def square(number):
    return number * number

print(square(5) + 10)             # a returned value can be used anywhere

def test():
    print("Start")
    return 10
    print("End")                  # never runs: return ends the function

test()
```

## Scope

`04_scope.py`

Scope: a variable created inside a function is local and dies with it.
A function can read a global variable. Prefer passing values in and returning them out.


```python
name = "Uma"

def greet():
    print("Hello", name)          # reads the global

greet()

def make_message():
    message = "Hello"             # local
    return message                # the way to get it out

print(make_message())

def calculate():
    result = 10 + 20
    return result

calculate()
# print(result)                   # NameError: result was local
```

## Defaults keywords

`05_defaults_keywords.py`

Default arguments fill a gap. Keyword arguments match by name, not position.


```python
def greet(name="Student"):
    print("Hello", name)

greet()
greet("Prince")

def power(number, exponent=2):
    return number ** exponent

print(power(5), power(5, 3))

def introduce(name, age):
    print(name, age)

introduce(age=20, name="Vinit")   # by name: order does not matter

def greet_with(name, message="Welcome"):
    print(message, name)

greet_with("Vinit")
greet_with("Vinit", "Good morning")
greet_with(name="Vinit", message="Good evening")
```

## Validation

Validation helpers: small functions that answer yes or no.

## Lambda

`07_lambda.py`

lambda: a one-expression function with no name of its own.
Use it for small, throw-away operations; def for anything longer.


```python
def square(x):
    return x * x

square = lambda x: x * x
print(square(5))

add = lambda a, b: a + b
print(add(10, 20))

double = lambda x: x * 2
print(double(3) + double(4))
```

## Score

`08_score.py`

Small functions, one job each: one computes, one decides, the main program connects them.


```python
def calculate_average(score1, score2, score3):
    total = score1 + score2 + score3
    return total / 3

def check_result(average):
    if average >= 40:
        return "Pass"
    return "Fail"

average = calculate_average(80, 90, 70)
result = check_result(average)
print("Average:", average)
print("Result:", result)
```

```python
def greet(name="Student"):
    print("Hello", name)

def check_even(number):
    if number % 2 == 0:
        return "Even"
    return "Odd"

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

greet()
print(check_even(10))
print(add(10, 5), subtract(10, 5), multiply(10, 5), divide(10, 5))
```

## Bill

`11_bill.py`

Mini challenge: a bill with a default tax rate in whole percent.
Money stays a whole number of rupees.


```python
def calculate_bill(price, quantity, tax=5):
    subtotal = price * quantity
    return subtotal + subtotal * tax // 100

print(calculate_bill(100, 2))
print(calculate_bill(100, 2, tax=10))
```

## Paisapay v5

`paisapay_v5.py`

PaisaPay v5: named blocks
Built in Class 5. The v4 menu loop stays. Every job is a function:
two checks, fee, cashback, mode name, show_balance, add_money, ask_pin, pay.
Fee, cashback and payment modes from v3 are back. Money is whole rupees.


```python
print("Welcome to PaisaPay")

saved_pin = "4321"

def is_valid_amount(amount):
    return amount > 0 and amount <= 100000

def is_valid_pin(pin):
    return pin == saved_pin

def fee_for(mode):
    if mode == "card":
        return 5
    elif mode == "netbanking":
        return 2
    return 0

def cashback_for(amount):
    if amount >= 5000:
        return 100
    elif amount >= 2000:
        return 50
    elif amount >= 500:
        return 10
    return 0

def mode_name(mode):
    match mode:
        case "upi":
            return "UPI"
        case "card":
            return "Card"
        case "netbanking":
            return "Net banking"
        case _:
            return "Unknown mode"

def show_balance(balance):
    print("Balance: ₹", balance)

def add_money(balance, amount):
    if not is_valid_amount(amount):
        print("Invalid amount. Nothing added.")
        return balance
    print("Added ₹", amount)
    return balance + amount

def ask_pin():
    for attempt in range(3):
        pin = input("UPI PIN: ")
        if is_valid_pin(pin):
            return True
        print("Wrong PIN.", 2 - attempt, "attempts left")
    return False

def pay(balance, payee, amount, mode="upi"):
    if not is_valid_amount(amount):
        print("Invalid amount. Payment refused.")
        return balance
    if mode_name(mode) == "Unknown mode":
        print("Unknown payment mode. Payment refused.")
        return balance
    fee = fee_for(mode)
    if amount + fee > balance:
        print("Insufficient balance. Payment refused.")
        return balance
    cashback = cashback_for(amount)
    print("Paid ₹", amount, "to", payee, "via", mode_name(mode))
    print("Fee ₹", fee, "| Cashback ₹", cashback)
    return balance - amount - fee + cashback

balance = 1000
choice = ""
while choice != "4":
    print()
    print("1. Pay   2. Balance   3. Add money   4. Exit")
    choice = input("Choose: ")

    if choice == "1":
        payee = input("Pay to: ")
        amount = int(input("Amount in rupees: "))
        mode = input("Mode (upi / card / netbanking): ")
        if ask_pin():
            balance = pay(balance, payee, amount, mode)
        else:
            print("Three wrong PINs. Payment refused.")

    elif choice == "2":
        show_balance(balance)

    elif choice == "3":
        amount = int(input("Amount to add: "))
        balance = add_money(balance, amount)

    elif choice != "4":
        print("Unknown choice. Try again.")

print("Thank you for using PaisaPay")

# What v5 cannot do yet:
#   - remember payments: the mini statement needs a list that grows (Class 6)
#   - clean up typed input or show paise (strings and formatting, Class 7)
#   - validate a UPI ID or PAN properly (regex, Class 11)
```
