## Regex at work

`01_regex_at_work.py`

Regex at work: last class's tools on real Indian data.
fullmatch for a whole field, search for one piece, findall for every piece.


```python
import re

# a form field must be exactly six digits: fullmatch, not search
for pin in ["560001", "5600011", "56001"]:
    print(pin, re.fullmatch(r"\d{6}", pin) is not None, re.search(r"\d{6}", pin) is not None)

# one piece out of a message
sms = "Your OTP is 482913. Do not share it."
print(re.search(r"\d{6}", sms).group())

# every piece: findall gives strings, so convert before adding
text = "Rs.600 to Sumit, Rs.1500 to Vamsi"
amounts = re.findall(r"Rs\.(\d+)", text)
print(amounts)
print(sum(int(a) for a in amounts))

# a PAN typed in lowercase fails until it is cleaned
pan = "abcde1234f"
print(re.fullmatch(r"[A-Z]{5}\d{4}[A-Z]", pan) is not None)
print(re.fullmatch(r"[A-Z]{5}\d{4}[A-Z]", pan.upper()) is not None)
```

## Inheritance scenarios

`02_inheritance_scenarios.py`

Inheritance scenarios: which code runs, and why.
1. A parent method calls a method the child overrides: the child's version runs


```python
class Account:
    def fee_for(self, mode):
        return 5 if mode == "card" else 0

    def total(self, amount, mode):
        return amount + self.fee_for(mode)    # self is whatever object called total

class PremiumAccount(Account):
    def fee_for(self, mode):
        return 0

print(Account().total(1000, "card"))          # 1005
print(PremiumAccount().total(1000, "card"))   # 1000

# 2. A child class variable hides the parent's
class Deposit:
    rate = 4

class FixedDeposit(Deposit):
    rate = 7

print(Deposit.rate, FixedDeposit.rate, FixedDeposit().rate)   # 4 7 7

# 3. super() works in any method, not only __init__
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def describe(self):
        return f"{self.name} earns {self.salary}"

class Manager(Employee):
    def describe(self):
        return super().describe() + " and leads a team"

print(Manager("Chitransh", 90000).describe())

# 4. super().__init__ needs every argument the parent asks for
class Intern(Employee):
    def __init__(self, name):
        super().__init__(name)                # Employee also needs salary

try:
    Intern("Karthikeyan")
except TypeError as e:
    print("TypeError:", e)
```

## Oop recap

`03_oop_recap.py`

OOP recap: one blueprint, class variables and instance variables,
encapsulation, and self choosing the right object.


```python
class Account:
    bank = "PaisaPay"                 # class variable: shared by every object

    def __init__(self, owner, balance):
        self.owner = owner            # instance variables: each object's own
        self.__balance = balance      # private: change it only through methods

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

rajani = Account("Rajani", 1000)
sumit = Account("Sumit", 2500)

Account.bank = "PaisaPay Plus"        # changed on the class: everyone sees it
print(rajani.bank, sumit.bank)

rajani.bank = "OtherPay"              # changed on one object: a new instance variable
print(rajani.bank, sumit.bank, Account.bank)

rajani.deposit(500)                   # self is rajani, so only her balance changes
print(rajani.get_balance(), sumit.get_balance())

try:
    print(rajani.__balance)
except AttributeError as e:
    print("AttributeError:", e)
```
