## Condition

`01_condition.py`

A condition is a question whose answer is True or False.


```python
balance = 500
amount = 200
print(balance >= amount)   # True

# Boundaries matter: > is strictly greater
amount = 2000
limit = 2000
print(amount > limit)      # False
print(amount >= limit)     # True
```

## If

`02_if.py`

if: run the indented block only when the condition is True.


```python
balance = 1000
amount = 400

if balance >= amount:
    print("Payment successful")

# When the condition is False, the block is skipped and nothing prints.
balance = 300
if balance >= amount:
    print("Payment successful")
print("Thank you for using PaisaPay")   # not indented: outside the if, runs every time

# Two errors to try, one at a time:
#   if balance = amount:      -> SyntaxError  (= assigns, == compares)
#   if balance >= amount:
#   print("x")                -> IndentationError (the block must be indented)
```

## Else

`03_else.py`

else: everything else. Exactly one of the two blocks runs.


```python
balance = 300
amount = 800

if balance >= amount:
    print("Payment successful")
else:
    print("Insufficient balance")

# Equal counts as enough: 500 >= 500 is True
balance = 500
amount = 500
if balance >= amount:
    print("Payment successful")
else:
    print("Insufficient balance")
```

## Elif cashback

`04_elif_cashback.py`

elif: more than two outcomes. Top to bottom, the first True wins.


```python
amount = int(input("Amount in rupees: "))

if amount >= 5000:
    cashback = 100
elif amount >= 2000:
    cashback = 50
elif amount >= 500:
    cashback = 10
else:
    cashback = 0

print("Cashback: ₹", cashback)
# Try 2500 (50), 5000 (100), 499 (0)
```

## Order matters

`05_order_matters.py`

The order of conditions matters: the first True wins.


```python
amount = 8000

# Bug: the broad condition comes first, so ₹8000 gets ₹10
if amount >= 500:
    print("Cashback ₹10")
elif amount >= 5000:
    print("Cashback ₹100")

# Fixed: most specific first
if amount >= 5000:
    print("Cashback ₹100")
elif amount >= 500:
    print("Cashback ₹10")

# Not a fix: two separate ifs are two independent questions, both print
if amount >= 5000:
    print("Cashback ₹100")
if amount >= 500:
    print("Cashback ₹10")
```

## Match case

`06_match_case.py`

match / case: one variable against exact values. Python 3.10 or newer.


```python
mode = input("Payment mode: ")

match mode:
    case "upi":
        print("Opening UPI app...")
    case "card":
        print("Enter card PIN")
    case "netbanking":
        print("Redirecting to your bank")
    case _:
        print("Unknown mode")     # the wildcard: anything left over

# Try "upi", then "UPI": a case is an exact, case-sensitive match.
```

## And or not

`07_and_or_not.py`

Conditions built with and, or, not


```python
balance = 1000
amount = 400
pin_correct = True

if balance >= amount and pin_correct:      # both must be True
    print("Payment successful")
else:
    print("Payment failed")

has_upi_pin = False
has_fingerprint = True
if has_upi_pin or has_fingerprint:         # at least one must be True
    print("Authorised")

is_frozen = False
if not is_frozen:                          # not flips the value
    print("Account active. Proceed.")

# A classic trap: this is always True, because "card" alone counts as True
mode = "cash"
if mode == "upi" or "card":
    print("Digital payment")
# Write it out instead:
if mode == "upi" or mode == "card":
    print("Digital payment")
else:
    print("Cash payment")
```

## Nested

`08_nested.py`

Nested: the inner question is asked only when the outer answer is True.


```python
pin_correct = True
balance = 1000
amount = 400

if pin_correct:
    if balance >= amount:
        print("Payment successful")
    else:
        print("Insufficient balance")
else:
    print("Wrong PIN")

# The same check with and: shorter, but one failure message for everything
if pin_correct and balance >= amount:
    print("Payment successful")

# An else belongs to the if at the SAME indentation.
# Delete the inner else above and, with balance 300, nothing prints at all.
```

## Ternary

`10_ternary.py`

The ternary operator: a compact if / else that gives a value.


```python
balance = 1000
amount = 400

if balance >= amount:
    status = "SUCCESS"
else:
    status = "FAILED"
print(status)

status = "SUCCESS" if balance >= amount else "FAILED"   # value_if_true if condition else value_if_false
print(status)

# Readable only for a simple two-way choice. Three or more outcomes: use if / elif / else.
amount = 1000
fee = 0 if amount < 1000 else 5 if amount < 10000 else 10   # hard to read; prints 5
print(fee)
```

## Paisapay v3

`paisapay_v3.py`

PaisaPay v3: the app can say no
Built in Class 3. Adds decisions to v2: wrong PIN, unknown mode and
insufficient balance are refused, cashback comes in tiers, and the
payment mode is matched against exact values.


```python
print("Welcome to PaisaPay")

balance = 1000
saved_pin = "4321"

name = input("Your name: ")
payee = input("Pay to: ")
amount = int(input("Amount in rupees: "))
mode = input("Mode (upi / card / netbanking): ")
pin = input("UPI PIN: ")

fee = 2
total = amount + fee

if amount >= 5000:
    cashback = 100
elif amount >= 2000:
    cashback = 50
elif amount >= 500:
    cashback = 10
else:
    cashback = 0

match mode:
    case "upi":
        mode_text = "UPI"
    case "card":
        mode_text = "Card"
    case "netbanking":
        mode_text = "Net banking"
    case _:
        mode_text = "Unknown mode"

print("----- Receipt -----")
if pin != saved_pin:
    print("Wrong PIN. Payment refused.")
elif mode_text == "Unknown mode":
    print("Unknown payment mode. Payment refused.")
elif total > balance:
    print("Insufficient balance. Payment refused.")
else:
    balance = balance - total + cashback
    print("Paid: ₹", amount, "to", payee, "via", mode_text)
    print("Cashback: ₹", cashback)
    print("Balance left: ₹", balance)

# What v3 cannot do yet:
#   - take a second payment: it handles one and stops (needs loops, Class 4)
#   - give the user another try after a wrong PIN (loops again)
```
